# Random forest classifier for written numbers

```python
from sklearn.model_selection import train_test_split
from sklearn.model_selection import KFold, ParameterGrid
import numpy as np
import numpy as np
from RandomForest import RandomForest
import pandas as pd
import os
from RF_preprocessing import load_samples, preprocess_samples
import pickle
```

## Data preprocessing

```python
samples, labels = load_samples("../training_data")
X, y = preprocess_samples(samples), np.array(labels)
print(X.shape,y.shape)
```

## Model

```python
def accuracy(y_true, y_pred):
    accuracy = np.sum(y_true == y_pred) / len(y_true)
    return accuracy
```

### K-fold cross validation for models generalization

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)

cv = KFold(n_splits=5, shuffle=True, random_state=0)

folds = []

for i, (train_idx, val_idx) in enumerate(cv.split(X_train), 1):
    print(f"Fold {i}: train={len(train_idx)} val={len(val_idx)}")
    folds.append((train_idx, val_idx))

accs = []
for train_idx, val_idx in folds:
    clf = RandomForest(n_trees=30, max_depth=30, min_samples_split=2, n_feature=26)
    clf.fit(X_train[train_idx], y_train[train_idx])
    predictions = clf.predict(X_train[val_idx])
    accs.append(accuracy(y_train[val_idx], predictions))
```

```python
accs = pd.Series(accs)
print(accs)
print(accs.mean())
```

### Training a model with all the training data

```python

clf = RandomForest(n_trees=50, max_depth=30, min_samples_split=2, n_feature=26)
clf.fit(X_train, y_train)
predictions = clf.predict(X_test)

acc =  accuracy(y_test, predictions)
print(np.round(acc),3)
```

### Saving it for testing purposes

```python
# Saving the model as pkl file
trained_models_dir = os.path.join("..", "trained_models")
model_path = os.path.join(trained_models_dir, "RF_classifier.pkl")

with open(model_path, "wb") as f: pickle.dump(clf, f)
```
